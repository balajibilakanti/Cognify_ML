```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import folium
import geopandas as gpd
```

```python
df = pd.read_csv('Dataset.csv') 
```

```python
df.head(5)
```

## Scatter plot of restaurants in the co-ordinates

```python
if "Latitude" and "Longitude" in df.columns:
  plt.figure(figsize=(10, 6))
  plt.scatter(df["Longitude"], df["Latitude"])
  plt.xlabel("Longitude")
  plt.ylabel("Latitude")
  plt.title("Restaurant Distribution")
  plt.show()
else:
  print("Latitude and longitude columns not found.")
```

## Visualizing the distribution on a map

```python
rest_gdf = gpd.GeoDataFrame(df, geometry=gpd.points_from_xy(df['Longitude'], df['Latitude']))
```

```python
map_rest = folium.Map(location=[rest_gdf['geometry'].y.mean(), rest_gdf['geometry'].x.mean()], zoom_start=12) 
rest_gdf.crs = 'EPSG:4326'
folium.GeoJson(rest_gdf).add_to(map_rest)
```
